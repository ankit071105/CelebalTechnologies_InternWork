````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🚀 BONUS TOOL 3: Text Summarizer

def summarize(text: str) -> str:
    """Return a short summary by extracting the first sentence."""
    try:
        sentences = text.split('.')
        summary = sentences[0].strip()
        return summary if summary else "Nothing to summarize."
    except Exception:
        return "Error in summarization"


# 🚀 BONUS TOOL 4: Word Counter

def word_count(text: str) -> dict:
    """Count total words and unique words in text."""
    try:
        words = text.lower().split()
        return {
            "total_words": len(words),
            "unique_words": len(set(words))
        }
    except Exception:
        return {"total_words": 0, "unique_words": 0}
````

````python
# 🚀 BONUS: Logging Setup

import logging
import re

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s | %(levelname)s | %(message)s',
    datefmt='%H:%M:%S'
)
logger = logging.getLogger('AgentPipeline')

print("Logger initialized successfully.")
````

## Implement Agent Logic Below

 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
#  AGENT FUNCTION — FULLY IMPLEMENTED

def agent(query: str) -> dict:
    """
    Single-Agent Smart Assistant with conditional routing.

    Routing logic:
      - 'calculate' or math operators  → Calculator Tool
      - 'keywords' or 'extract'        → Keyword Extractor Tool
      - 'summarize'                    → Summarizer Tool  (Bonus)
      - 'count words' / 'word count'   → Word Counter Tool (Bonus)
      - anything else                  → General response

    Returns:
      dict with keys: type, result, tool_used, query
    """

    logger.info(f"Received query: '{query}'")
    query_lower = query.lower()

    try:
        # ── Route 1: Calculator ──────────────────────────────────────
        # Trigger: query has 'calculate', 'compute', or math operators
        if ("calculate" in query_lower or
            "compute" in query_lower or
            re.search(r'[\d]+\s*[\+\-\*\/\%\^]+\s*[\d]+', query)):

            logger.info("Route → Calculator Tool")

            # Extract the math expression from the query
            # e.g. "Calculate 20 + 5" → "20 + 5"
            expression = re.sub(
                r'(calculate|compute|what is|=\s*\?)\s*',
                '', query_lower
            ).strip()

            result = calculator(expression)
            logger.info(f"Calculator result: {result}")

            return {
                "type": "calculation",
                "result": result,
                "tool_used": "calculator",
                "query": query
            }

        # ── Route 2: Keyword Extractor ───────────────────────────────
        # Trigger: query has 'keywords', 'extract', 'key words'
        elif ("keywords" in query_lower or
              "extract" in query_lower or
              "key words" in query_lower):

            logger.info("Route → Keyword Extractor Tool")

            # Strip the routing trigger words to get just the text
            text = re.sub(
                r'(extract keywords from|keywords from|extract|keywords)',
                '', query_lower
            ).strip()

            keywords = extract_keywords(text)
            logger.info(f"Keywords extracted: {keywords}")

            return {
                "type": "keywords",
                "result": keywords,
                "tool_used": "keyword_extractor",
                "query": query
            }

        # ── Route 3: Summarizer (Bonus) ──────────────────────────────
        elif "summarize" in query_lower or "summary" in query_lower:

            logger.info("Route → Summarizer Tool")

            text = re.sub(r'(summarize|give me a summary of|summary of)', '', query_lower).strip()
            result = summarize(text)
            logger.info(f"Summary: {result}")

            return {
                "type": "summary",
                "result": result,
                "tool_used": "summarizer",
                "query": query
            }

        # ── Route 4: Word Counter (Bonus) ────────────────────────────
        elif "count words" in query_lower or "word count" in query_lower:

            logger.info("Route → Word Counter Tool")

            text = re.sub(r'(count words in|word count of|count words)', '', query_lower).strip()
            result = word_count(text)
            logger.info(f"Word count: {result}")

            return {
                "type": "word_count",
                "result": result,
                "tool_used": "word_counter",
                "query": query
            }

        # ── Route 5: General Response ────────────────────────────────
        else:
            logger.info("Route → General Response (no tool)")

            general_response = (
                f"I received your query: '{query}'. "
                "I can help with calculations, keyword extraction, "
                "summarization, and word counting. "
                "Please rephrase using trigger words like "
                "'calculate', 'keywords', 'summarize', or 'count words'."
            )

            return {
                "type": "general",
                "result": general_response,
                "tool_used": None,
                "query": query
            }

    except Exception as e:
        # ── Error Handler ────────────────────────────────────────────
        logger.error(f"Agent error: {str(e)}")
        return {
            "type": "error",
            "result": f"An error occurred: {str(e)}",
            "tool_used": None,
            "query": query
        }
````
